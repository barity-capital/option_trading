```python
from binance.client import Client
from datetime import datetime, timedelta
import re

# Initialize the Binance client
api_key = os.getenv('BINANCE_API_KEY')
api_secret = os.getenv('BINANCE_API_SECRET')
client = Client(api_key, api_secret)

# Define the target symbol and lookahead period
base_symbol = 'ETHUSDT'
target_date = datetime.now() + timedelta(days=30)

# Fetch all futures contracts for the symbol
futures_info = client.futures_exchange_info()
contracts = futures_info['symbols']

# Filter contracts to find the one closest to the target expiration date
closest_contract = None
min_diff = None

for contract in contracts:
    if contract['symbol'].startswith(base_symbol) and not contract['symbol'].endswith('PERP'):
        # Extract the expiration date from the contract symbol
        match = re.search(r'_(\d{6})$', contract['symbol'])
        if match:
            expiration_str = match.group(1)
            expiration_date = datetime.strptime(expiration_str, '%y%m%d')

            # Calculate the difference between the expiration date and the target date
            diff = abs((expiration_date - target_date).days)

            # Select the contract with the smallest difference
            if min_diff is None or diff < min_diff:
                min_diff = diff
                closest_contract = contract['symbol']

# Ensure a contract was found
if closest_contract:
    print(f"Selected contract: {closest_contract}")
    quantity = 0.008  # Adjust quantity as needed

    # Prompt the user for confirmation
    confirmation = input(f"Do you want to place a market order for {quantity} of {closest_contract}? (yes/no): ")

    if confirmation.lower() == 'yes':
        # Place the order for the selected contract
        try:
            order = client.futures_create_order(
                symbol=closest_contract,
                side=SIDE_BUY,
                type=ORDER_TYPE_MARKET,
                quantity=quantity
            )
            print(f"Order placed: {order}")
        except Exception as e:
            print(f"An error occurred: {e}")
    else:
        print("Order cancelled.")
else:
    print("No suitable futures contract found.")
```

```python
import requests
from binance.client import Client
from datetime import datetime, timedelta
import re

# Initialize the Binance client
api_key = os.getenv('BINANCE_API_KEY')
api_secret = os.getenv('BINANCE_API_SECRET')
client = Client(api_key, api_secret)

# Fetch options symbols manually from Binance's public API
def fetch_options_list(underlying="ETH"):
    url = f'https://eapi.binance.com/eapi/v1/order'
    response = requests.get(url)
    return response.json()

# Define the target symbol and lookahead period
base_symbol = 'ETH'
target_date = datetime.now() + timedelta(days=30)

# Fetch all options contracts for the underlying asset (ETH in this case)
options_info = fetch_options_list(base_symbol)
contracts = options_info.get('optionPairList', [])

# Filter contracts to find the one closest to the target expiration date
closest_contract = None
min_diff = None

for contract in contracts:
    for symbol in contract['options']:
        # Extract the expiration date from the contract symbol
        match = re.search(r'(\d{6})', symbol['symbol'])
        if match:
            expiration_str = match.group(1)
            expiration_date = datetime.strptime(expiration_str, '%y%m%d')

            # Calculate the difference between the expiration date and the target date
            diff = abs((expiration_date - target_date).days)

            # Select the contract with the smallest difference
            if min_diff is None or diff < min_diff:
                min_diff = diff
                closest_contract = symbol['symbol']

# Ensure a contract was found
if closest_contract:
    print(f"Selected contract: {closest_contract}")
    quantity = 1  # Adjust quantity as needed

    # Prompt the user for confirmation
    confirmation = input(f"Do you want to place a market order for {quantity} contract(s) of {closest_contract}? (yes/no): ")

    if confirmation.lower() == 'yes':
        # Place the order for the selected contract
        try:
            order = client.options_create_order(
                symbol=closest_contract,
                side=Client.SIDE_BUY,
                type=Client.ORDER_TYPE_MARKET,
                quantity=quantity
            )
            print(f"Order placed: {order}")
        except Exception as e:
            print(f"An error occurred: {e}")
    else:
        print("Order cancelled.")
else:
    print("No suitable options contract found.")
```
